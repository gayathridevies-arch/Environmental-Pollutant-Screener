In Python, dictionaries store information in pairs: a Key (the label) and a Value (the data attached to it).

For example, in search_patterns dictionary:

Key: "Has_Halogen"

Value: "[#6]-[Cl,Br,I]" (the SMARTS pattern)

Here is exactly what .keys() does:

1. .keys() extracts only the labels
When we call search_patterns.keys(), we are telling Python: "Look inside this dictionary and give only the keys (the column names), and ignore the SMARTS string values."

It pulls out: ['Has_Halogen', 'Has_Nitroaromatic', 'Has_Phenolic', 'Has_Aromatic_Amine']



2. In Pandas, .any() checks if at least one value in that row is True.

To put it into perspective:

.any(axis=1): Flags the row if any of the hazard columns are True (meaning: "Did this chemical trigger at least one alert?").

.all(axis=1): Would flag the row only if all of the columns are True (meaning: "Did this chemical trigger every single alert on the list?")
